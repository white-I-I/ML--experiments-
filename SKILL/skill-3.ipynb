```python
# ============================================================
# PROJECT: FEATURE ENGINEERING AND EXPLORATORY DATA ANALYSIS
# DATASET: WineQT.csv
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

file_path = "/content/drive/MyDrive/Colab/WineQT.csv"

df = pd.read_csv(file_path)

print("\nDataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())


# ============================================================
# 3. DATA UNDERSTANDING
# ============================================================

print("\n" + "=" * 60)
print("DATA UNDERSTANDING")
print("=" * 60)

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types and Non-Null Values:")
df.info()


# ============================================================
# 4. STATISTICAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)

display(df.describe())


# ============================================================
# 5. MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUE ANALYSIS")
print("=" * 60)

missing_values = df.isnull().sum()

print(missing_values)


# ============================================================
# 6. DUPLICATE VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("DUPLICATE VALUE ANALYSIS")
print("=" * 60)

duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)


# ============================================================
# 7. DATA CLEANING
# ============================================================

print("\n" + "=" * 60)
print("DATA CLEANING")
print("=" * 60)

# Remove duplicate rows
df = df.drop_duplicates()

# Select numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

# Fill missing values with median
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Data cleaning completed!")

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nDataset Shape After Cleaning:")
print(df.shape)


# ============================================================
# 8. TARGET VARIABLE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("WINE QUALITY DISTRIBUTION")
print("=" * 60)

print(df["quality"].value_counts().sort_index())


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="quality"
)

plt.title("Distribution of Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Number of Wines")

plt.show()


# ============================================================
# 9. DISTRIBUTION OF NUMERICAL FEATURES
# ============================================================

print("\n" + "=" * 60)
print("FEATURE DISTRIBUTION")
print("=" * 60)

numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns].hist(
    figsize=(16, 12),
    bins=20,
    edgecolor="black"
)

plt.suptitle(
    "Distribution of Numerical Features",
    fontsize=16
)

plt.tight_layout()

plt.show()


# ============================================================
# 10. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("CORRELATION ANALYSIS")
print("=" * 60)

correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap of Wine Features")

plt.show()


# ============================================================
# 11. CORRELATION WITH WINE QUALITY
# ============================================================

print("\nCorrelation of Features with Wine Quality:\n")

quality_correlation = correlation_matrix["quality"].sort_values(
    ascending=False
)

print(quality_correlation)


# ============================================================
# 12. OUTLIER DETECTION USING BOXPLOTS
# ============================================================

print("\n" + "=" * 60)
print("OUTLIER DETECTION")
print("=" * 60)

for column in numeric_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(
        x=df[column]
    )

    plt.title(
        f"Boxplot of {column}"
    )

    plt.show()


# ============================================================
# 13. RELATIONSHIP BETWEEN ALCOHOL AND WINE QUALITY
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="quality",
    y="alcohol"
)

plt.title("Alcohol Content vs Wine Quality")

plt.xlabel("Wine Quality")
plt.ylabel("Alcohol Content")

plt.show()


# ============================================================
# 14. FEATURE ENGINEERING
# ============================================================

print("\n" + "=" * 60)
print("FEATURE ENGINEERING")
print("=" * 60)


# ------------------------------------------------------------
# FEATURE 1: TOTAL ACIDITY
# ------------------------------------------------------------

df["total_acidity"] = (
    df["fixed acidity"]
    + df["volatile acidity"]
    + df["citric acid"]
)

print("Feature created: total_acidity")


# ------------------------------------------------------------
# FEATURE 2: SULPHATES TO ALCOHOL RATIO
# ------------------------------------------------------------

df["sulphates_alcohol_ratio"] = (
    df["sulphates"]
    / df["alcohol"]
)

print("Feature created: sulphates_alcohol_ratio")


# ------------------------------------------------------------
# FEATURE 3: SUGAR TO ACIDITY RATIO
# ------------------------------------------------------------

df["sugar_acidity_ratio"] = (
    df["residual sugar"]
    / df["total_acidity"]
)

print("Feature created: sugar_acidity_ratio")


# ------------------------------------------------------------
# FEATURE 4: QUALITY CATEGORY
# ------------------------------------------------------------

df["quality_category"] = np.where(
    df["quality"] >= 7,
    "Good",
    "Average"
)

print("Feature created: quality_category")


# ============================================================
# 15. VIEW FEATURE ENGINEERED DATA
# ============================================================

print("\n" + "=" * 60)
print("NEWLY CREATED FEATURES")
print("=" * 60)

display(
    df[
        [
            "total_acidity",
            "sulphates_alcohol_ratio",
            "sugar_acidity_ratio",
            "quality_category"
        ]
    ].head()
)


# ============================================================
# 16. QUALITY CATEGORY DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="quality_category"
)

plt.title("Wine Quality Category Distribution")

plt.xlabel("Quality Category")
plt.ylabel("Number of Wines")

plt.show()


# ============================================================
# 17. ALCOHOL VS QUALITY CATEGORY
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="quality_category",
    y="alcohol"
)

plt.title("Alcohol Content by Quality Category")

plt.xlabel("Quality Category")
plt.ylabel("Alcohol Content")

plt.show()


# ============================================================
# 18. TOTAL ACIDITY VS ALCOHOL
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="total_acidity",
    y="alcohol",
    hue="quality_category"
)

plt.title("Total Acidity vs Alcohol")

plt.xlabel("Total Acidity")
plt.ylabel("Alcohol")

plt.show()


# ============================================================
# 19. FINAL DATASET INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("FINAL FEATURE ENGINEERED DATASET")
print("=" * 60)

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nFinal Column Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows of Final Dataset:")
display(df.head())


# ============================================================
# 20. SAVE THE FINAL DATASET TO GOOGLE DRIVE
# ============================================================

output_path = (
    "/content/drive/MyDrive/Colab/"
    "WineQT_Feature_Engineered.csv"
)

df.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 60)

print("\nFinal dataset saved at:")
print(output_path)
```
